```python

def read_rainfall_recorded(file_name):
    months = []
    rainfall_values = []
    try:
      with open(file_name, "r") as file:
        for line in file:
            month, rainfall = line.strip().split(",") #splitting the line into month and rainfall values
            months.append(month)
            try: #accounting for invalid entries
                rainfall_values.append(float(rainfall))
            except ValueError:
                print(f"Invalid rainfall value for {month}: {rainfall}")
        file.close()
        print("Rainfall values and months have been read successfully")
    except FileNotFoundError:
        print(f"File {file_name} not found.")
    return months, rainfall_values

def calculate_total_rainfall(rainfall_values):
    return sum(rainfall_values)

def calculate_average_rainfall(rainfall_values):
    return sum(rainfall_values) / len(rainfall_values) 

def find_highest_rainfall(rainfall_values , months):
    highest_rainfall = max(rainfall_values) #identifying highest rainfall value 
    highest_month = months[rainfall_values.index(highest_rainfall)] # matching rainfall value to month 
    return highest_month, highest_rainfall

def find_lowest_rainfall(rainfall_values , months):
    lowest_rainfall = min(rainfall_values) # searching for the lowest value 
    lowest_month = months[rainfall_values.index(lowest_rainfall)] # matching lowest rqinfall value to month
    return lowest_month, lowest_rainfall

def identify_drought_months(rainfall_values, months):
    drought_values = list(filter(lambda x: x < 25 , rainfall_values)) #using lambda and filter to identify months with rainfall below 25mm
    drought_months = list(months[rainfall_values.index(rain)] for rain in drought_values) # matching the drought values to their respective months 
    return drought_months

def identify_seasons(months):

    #categorising months into seasons

    summer_months = []
    autumn_months = []
    winter_months = []
    spring_months = []
  
    
    for month in months:
        if month in ["December", "January", "February"]:
            summer_months.append(month)
        elif month in ["March", "April", "May"]:
            autumn_months.append(month)
        elif month in ["June", "July", "August"]:
            winter_months.append(month)
        elif month in ["September", "October", "November"]:
            spring_months.append(month)
        
    return summer_months, autumn_months, winter_months, spring_months

def compare_seasons(season1, season2, rainfall_values, months):
    season1_rainfall = sum(rainfall_values[months.index(month)] for month in season1) #gettoing the total rainfall for season 1
    season2_rainfall = sum(rainfall_values[months.index(month)] for month in season2) #getting the total rainfall for season 2

    if season1_rainfall > season2_rainfall: #comparing the total rainfall for both seasons and returning the result
        return f"{season1} was wetter than {season2}."
    elif season1_rainfall < season2_rainfall:
        return f"{season2} was wetter than {season1}."
    
    if season1_rainfall > season2_rainfall:
        return f"{season1} was wetter than {season2}."
    elif season1_rainfall < season2_rainfall:
        return f"{season2} was wetter than {season1}."
    else:
        return f"{season1} and {season2} had the same amount of rainfall."

def write_summary_report(file_name, total_rainfall, average_rainfall, highest_month, highest_rainfall, lowest_month, lowest_rainfall, drought_months, season_comparison):
    try:
       with open(file_name, "w") as file: #writing the summary report to a file
        file.write(f"Total Rainfall: {total_rainfall} mm\n")
        file.write(f"Average Rainfall: {average_rainfall} mm\n")
        file.write(f"Highest Rainfall: {highest_rainfall} mm in {highest_month}\n")
        file.write(f"Lowest Rainfall: {lowest_rainfall} mm in {lowest_month}\n")
        file.write(f"Drought Months (below 25mm): {drought_months}\n")
        file.write(f"Season Comparison: {season_comparison}\n")
        print("Summary report written to", file_name , "successfully")
    except Exception as file_not_found_error:
        print(f"Could not write to file , {file_name} as file could not be found ")

def season_comparison_menu():
    print("Which seasons would you like to compare?")
    print("1. Summer and Autumn")
    print("2. Winter and Spring")
    print("3. Summer and Winter")
    print("4. Autumn and Spring")
    season_choice = input("Enter your choice (1-4): ")
    if season_choice == "1":
        return compare_seasons(summer_months, autumn_months, rainfall_recorded, months)
    elif season_choice == "2":
        return compare_seasons(winter_months, spring_months, rainfall_recorded, months)
    elif season_choice == "3":
        return compare_seasons(summer_months, winter_months, rainfall_recorded, months)
    elif season_choice == "4":
        return compare_seasons(autumn_months, spring_months, rainfall_recorded, months)
    else:
        print("Invalid choice. Please select a valid option.")
        return None

def menu(): #menu to allow user to select wanted options
   print(" Team seven presents \n == RAINFALL DATA ANALYSER ==") 
   while True:
       print("\nMenu:")
       print("1. Read rainfall data from file")
       print("2. Calculate total rainfall")
       print("3. Calculate average rainfall")
       print("4. Find highest rainfall month")
       print("5. Find lowest rainfall month")
       print("6. Identify drought months")
       print("7. Compare two seasons")
       print("8. Write summary report to file")
       print("9. Exit")

       choice = input("Enter your choice (1-9): ")

       if choice == "1":
           print("\n \n")
           print(f"Months : {months} Rainfall Recorded : {rainfall_recorded}")
           print(f"Summer Months: {summer_months} \n, Autumn Months: {autumn_months} \n , Winter Months: {winter_months} \n, Spring Months: {spring_months} \n")
                 
       elif choice == "2":
           print("\n \n")
           total_rainfall = calculate_total_rainfall(rainfall_recorded)
           print("Total Rainfall:", total_rainfall, "mm")
       elif choice == "3":
           print("\n \n")
           average_rainfall = calculate_average_rainfall(rainfall_recorded)
           print("Average Rainfall:", average_rainfall, "mm")    
       elif choice == "4":
           print("\n \n")
           highest_month, highest_rainfall = find_highest_rainfall(rainfall_recorded , months)
           print("Highest Rainfall:", highest_rainfall, "mm in", highest_month)
       elif choice == "5":
           print("\n \n")
           lowest_month, lowest_rainfall = find_lowest_rainfall(rainfall_recorded , months)
           print("Lowest Rainfall:", lowest_rainfall, "mm in", lowest_month)
       elif choice == "6":
           print("\n \n")
           drought_months = identify_drought_months(rainfall_recorded, months)
           print("Drought Months (below 25mm):", drought_months)
       elif choice == "7":
           print("\n \n")
           season_comp_result = season_comparison_menu()
           print("Season Comparison Result:", season_comp_result)
       elif choice == "8":
           print("\n \n")
           file_name = input("Enter the file name to write the summary report: ")
           total_rainfall = calculate_total_rainfall(rainfall_recorded)
           average_rainfall = calculate_average_rainfall(rainfall_recorded)
           highest_month, highest_rainfall = find_highest_rainfall(rainfall_recorded, months)
           lowest_month, lowest_rainfall = find_lowest_rainfall(rainfall_recorded, months)
           drought_months = identify_drought_months(rainfall_recorded, months)
           season_comp_result = season_comparison_menu()
           write_summary_report(file_name, total_rainfall, average_rainfall, highest_month, highest_rainfall, lowest_month, lowest_rainfall, drought_months, season_comp_result)
       elif choice == "9":
           print(" Thank you for using the Rainfall Data Analyser")
           break
       else:
           print("Invalid choice. Please select a valid option.")


print("Members : \n Desire Myambo : 260449 \n Mraia Michael : 262366 \n Lloyd Sabata : 262088 \n Isaac Masaja : 262377 \n Kupakwashe Mandigo : 260136 \n Christopher Kanengoni : 260655 \n Tadiwanashe Mugiya : 262207 \n ")
months , rainfall_recorded = read_rainfall_recorded("rainfall2.txt")
summer_months, autumn_months, winter_months, spring_months = identify_seasons(months)
menu()




```
